```python
import pandas as pd
import json

results = "results_tdr12days_062026/results/capacity.csv"
locations_path = "system/locations_4.json"
```

```python
df = pd.read_csv(results, usecols=['zone', 'resource_type', 'new_capacity'])
df = df.iloc[4964:5109]

with open(locations_path) as f:
    locations = json.load(f)['locations']

max_new_capacity = {
    loc['id']: loc['constraints']['MaxNewCapacityConstraint']['value']
    for loc in locations
    if 'MaxNewCapacityConstraint' in loc.get('constraints', {})
}

df['max_new_capacity'] = df['zone'].map(max_new_capacity)
df
```
